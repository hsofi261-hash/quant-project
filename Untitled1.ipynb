```python
import pandas as pd
import numpy as np
import yfinance as yf

# ==========================================
# STEP 1: SOURCE THE DATA
# ==========================================
print("--- 1. DOWNLOADING DATA ---")
# ^NSEI is the Yahoo Finance ticker for NIFTY 50
df = yf.download("^NSEI", start="2010-01-01", end="2025-12-31")

# Clean multi-index columns if present in newer yfinance versions
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df = df[['Open', 'High', 'Low', 'Close']].dropna()
print(f"Data downloaded successfully. Shape: {df.shape}")


# ==========================================
# STEP 2: VALIDATE & CLEAN DATA
# ==========================================
print("\n--- 2. DATA VALIDATION CHECKS ---")
# 2a. Check duplicate dates
duplicates = df.index.duplicated().sum()
print(f"Duplicate dates found: {duplicates}")

# 2b. Check sorting
is_sorted = df.index.is_monotonic_increasing
print(f"Data properly sorted chronologically: {is_sorted}")
if not is_sorted:
    df = df.sort_index()

# 2c. Check invalid OHLC relationships (e.g. High < Low)
invalid_ohlc = (df['High'] < df['Low']).sum()
print(f"Invalid OHLC observations (High < Low): {invalid_ohlc}")

# 2d. Check missing / null values
null_count = df.isnull().sum().sum()
print(f"Missing values (NaNs): {null_count}")


# ==========================================
# STEP 3: EVENT DETECTION ENGINE (CONFIGURABLE)
# ==========================================
def analyze_events(data, threshold=-0.015, holding_days=5):
    """
    Detects market fall events and calculates realistic forward returns.

    Parameters:
    - data: DataFrame with Open and Close prices
    - threshold: Daily drop threshold (e.g., -0.015 for -1.5%)
    - holding_days: Number of days to hold position (H)
    """
    df_engine = data.copy()

    # Calculate daily percentage return
    df_engine['Daily_Return'] = df_engine['Close'].pct_change()

    # Identify event trigger (True if drop <= threshold)
    df_engine['Is_Event'] = df_engine['Daily_Return'] <= threshold

    events_list = []

    # Extract event indices
    event_indices = np.where(df_engine['Is_Event'])[0]

    for idx in event_indices:
        # Check boundary condition (ensure we have enough future days to measure exit)
        if idx + 1 + holding_days < len(df_engine):
            event_date = df_engine.index[idx]
            event_return = df_engine['Daily_Return'].iloc[idx]

            # Entry Price: Open price of next trading day (t+1)
            entry_price = df_engine['Open'].iloc[idx + 1]

            # Exit Price: Close price after holding_days (t + 1 + H)
            exit_price = df_engine['Close'].iloc[idx + 1 + holding_days]

            # Forward Return calculation
            fwd_return = (exit_price - entry_price) / entry_price

            events_list.append({
                'Event_Date': event_date,
                'Event_Return': event_return,
                'Entry_Date': df_engine.index[idx + 1],
                'Entry_Price': entry_price,
                'Exit_Date': df_engine.index[idx + 1 + holding_days],
                'Exit_Price': exit_price,
                'Forward_Return': fwd_return
            })

    events_df = pd.DataFrame(events_list)

    # Summary Statistics
    stats = {
        'Total Observations': len(events_df),
        'Mean Forward Return (%)': events_df['Forward_Return'].mean() * 100,
        'Median Forward Return (%)': events_df['Forward_Return'].median() * 100,
        'Win Rate (%)': (events_df['Forward_Return'] > 0).mean() * 100,
        'Volatility / Std Dev (%)': events_df['Forward_Return'].std() * 100
    }

    return events_df, pd.Series(stats)

# Run Engine for baseline configuration (-1.5% drop, 5 days hold)
print("\n--- 3. ENGINE RESULTS (BASELINE: -1.5% Drop, 5-Day Hold) ---")
events_df, stats_summary = analyze_events(df, threshold=-0.015, holding_days=5)

print(stats_summary.to_string())
print("\nFirst 3 Detected Events:")
print(events_df[['Event_Date', 'Event_Return', 'Entry_Price', 'Exit_Price', 'Forward_Return']].head(3))
```

```python
import scipy.stats as stats
import numpy as np
import pandas as pd

# ==========================================
# TASK 3: STATISTICAL EVIDENCE ENGINE
# ==========================================
print("--- TASK 3: STATISTICAL EVIDENCE EVALUATION ---")

# Extract the forward returns from Task 2 (in percentage)
fwd_returns = events_df['Forward_Return'].dropna()

# 1. PARAMETRIC HYPOTHESIS TESTING (One-Sample t-Test)
# Testing if mean return is significantly > 0
t_stat, p_value_two_tailed = stats.ttest_1samp(fwd_returns, popmean=0.0)
p_value_one_tailed = p_value_two_tailed / 2 if t_stat > 0 else 1.0 - (p_value_two_tailed / 2)

# 2. CONFIDENCE INTERVAL (95% Standard Parametric)
mean_ret = np.mean(fwd_returns)
sem = stats.sem(fwd_returns) # Standard error of the mean
ci_95 = stats.t.interval(0.95, df=len(fwd_returns)-1, loc=mean_ret, scale=sem)

# 3. BOOTSTRAP RESAMPLING (Non-parametric)
n_iterations = 10000
bootstrap_means = np.empty(n_iterations)

np.random.seed(42) # For reproducibility
for i in range(n_iterations):
    boot_sample = np.random.choice(fwd_returns, size=len(fwd_returns), replace=True)
    bootstrap_means[i] = np.mean(boot_sample)

boot_ci_lower = np.percentile(bootstrap_means, 2.5)
boot_ci_upper = np.percentile(bootstrap_means, 97.5)

# DISPLAY RESULTS
print(f"Sample Size (Events Count): {len(fwd_returns)}")
print(f"Mean Event Forward Return:  {mean_ret * 100:.3f}%")
print(f"Standard Error of Mean:    {sem * 100:.3f}%")
print("-" * 45)
print(f"t-Statistic:               {t_stat:.4f}")
print(f"p-Value (One-Tailed):      {p_value_one_tailed:.4f}")
print(f"95% Parametric CI:        [{ci_95[0]*100:.3f}%, {ci_95[1]*100:.3f}%]")
print(f"95% Bootstrap CI:         [{boot_ci_lower*100:.3f}%, {boot_ci_upper*100:.3f}%]")
print("-" * 45)

# Statistical Conclusion
if p_value_one_tailed < 0.05:
    print("VERDICT: Reject Null Hypothesis (H0). Returns are statistically significantly positive.")
else:
    print("VERDICT: Fail to Reject Null Hypothesis (H0). Insufficient statistical evidence of recovery.")
```

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# ==========================================
# TASK 4: BASELINE & ROBUSTNESS TESTING
# ==========================================
print("--- TASK 4: BASELINE & ROBUSTNESS ANALYSIS ---")

# 1. BASELINE COMPARISON
# Calculate 5-day forward return for EVERY trading day in the dataset
df['Baseline_5D_Fwd'] = (df['Close'].shift(-5) - df['Open'].shift(-1)) / df['Open'].shift(-1)
baseline_returns = df['Baseline_5D_Fwd'].dropna()

baseline_mean = baseline_returns.mean() * 100
baseline_median = baseline_returns.median() * 100
event_mean = events_df['Forward_Return'].mean() * 100

# Two-sample t-test: Event returns vs. Normal baseline returns
t_stat_base, p_val_base = stats.ttest_ind(events_df['Forward_Return'].dropna(), baseline_returns, equal_var=False)

print("1. BASELINE COMPARISON:")
print(f"Overall Market 5D Mean Return:   {baseline_mean:.3f}%")
print(f"Overall Market 5D Median Return: {baseline_median:.3f}%")
print(f"Event-Driven 5D Mean Return:     {event_mean:.3f}%")
print(f"Excess Return over Baseline:     {event_mean - baseline_mean:.3f}%")
print(f"Difference t-Stat vs Baseline:   {t_stat_base:.4f} (p-value: {p_val_base/2:.4f})")
print("-" * 50)


# 2. ROBUSTNESS / PARAMETER SENSITIVITY GRID
print("\n2. PARAMETER SENSITIVITY GRID (ROBUSTNESS CHECK):")

thresholds = [-0.010, -0.015, -0.020, -0.025] # -1.0%, -1.5%, -2.0%, -2.5% drops
holding_periods = [3, 5, 10]                   # 3-day, 5-day, 10-day holds

grid_results = []

for th in thresholds:
    for hp in holding_periods:
        ev_df, _ = analyze_events(df, threshold=th, holding_days=hp)
        fwd_ret = ev_df['Forward_Return'].dropna()

        if len(fwd_ret) > 0:
            m_ret = fwd_ret.mean() * 100
            w_rate = (fwd_ret > 0).mean() * 100
            t_s, p_v = stats.ttest_1samp(fwd_ret, 0.0)

            grid_results.append({
                'Threshold (%)': th * 100,
                'Holding (Days)': hp,
                'Obs Count': len(fwd_ret),
                'Mean Ret (%)': round(m_ret, 3),
                'Win Rate (%)': round(w_rate, 1),
                'p-Value': round(p_v / 2 if t_s > 0 else 1 - p_v/2, 4)
            })

robustness_df = pd.DataFrame(grid_results)
print(robustness_df.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats

# ==========================================
# TASK 5: OUT-OF-SAMPLE (OOS) VALIDATION
# ==========================================
print("--- TASK 5: OUT-OF-SAMPLE VALIDATION ---")

is_mask = (df.index >= '2010-01-01') & (df.index <= '2020-12-31')
oos_mask = (df.index >= '2021-01-01') & (df.index <= '2025-12-31')

df_is = df[is_mask]
df_oos = df[oos_mask]

events_is, stats_is = analyze_events(df_is, threshold=-0.015, holding_days=5)
events_oos, stats_oos = analyze_events(df_oos, threshold=-0.015, holding_days=5)

print(f"In-Sample (2010-2020)  : Obs = {len(events_is)}, Mean Ret = {events_is['Forward_Return'].mean()*100:.3f}%, Win Rate = {(events_is['Forward_Return']>0).mean()*100:.1f}%")
print(f"Out-of-Sample (2021-2025): Obs = {len(events_oos)}, Mean Ret = {events_oos['Forward_Return'].mean()*100:.3f}%, Win Rate = {(events_oos['Forward_Return']>0).mean()*100:.1f}%")
print("-" * 55)

# ==========================================
# TASK 6: CHALLENGE YOUR OWN RESULT (REGIME ANALYSIS)
# ==========================================
print("\n--- TASK 6: MARKET REGIME FALSIFICATION ---")
# Classify market regime using 200-day Simple Moving Average (SMA)
df['SMA200'] = df['Close'].rolling(window=200).mean()
events_df['SMA200_at_event'] = df.loc[events_df['Event_Date'], 'SMA200'].values
events_df['Event_Close'] = df.loc[events_df['Event_Date'], 'Close'].values
events_df['Regime'] = np.where(events_df['Event_Close'] >= events_df['SMA200_at_event'], 'Bull (Close >= SMA200)', 'Bear (Close < SMA200)')

regime_summary = events_df.groupby('Regime')['Forward_Return'].agg(['count', lambda x: x.mean()*100, lambda x: (x>0).mean()*100])
regime_summary.columns = ['Obs Count', 'Mean Ret (%)', 'Win Rate (%)']
print(regime_summary.to_string())
print("-" * 55)

# ==========================================
# TASK 7: SIMPLE EVENT-DRIVEN BACKTEST
# ==========================================
print("\n--- TASK 7: BACKTEST WITH FRICTION & LOCKOUT ---")

# Parameters
COST_PER_TRADE = 0.0010 # 10 bps round-trip friction
holding_days = 5
threshold = -0.015

df_bt = df.copy()
df_bt['Daily_Return'] = df_bt['Close'].pct_change()
df_bt['Is_Event'] = df_bt['Daily_Return'] <= threshold

trades = []
last_exit_idx = -1

for i in range(len(df_bt)):
    if df_bt['Is_Event'].iloc[i] and i > last_exit_idx:
        if i + 1 + holding_days < len(df_bt):
            entry_price = df_bt['Open'].iloc[i + 1]
            exit_price = df_bt['Close'].iloc[i + 1 + holding_days]

            raw_ret = (exit_price - entry_price) / entry_price
            net_ret = raw_ret - COST_PER_TRADE

            trades.append({
                'Entry_Date': df_bt.index[i + 1],
                'Exit_Date': df_bt.index[i + 1 + holding_days],
                'Raw_Return': raw_ret,
                'Net_Return': net_ret
            })
            last_exit_idx = i + 1 + holding_days

trades_df = pd.DataFrame(trades)
trades_df['Equity_Curve'] = (1 + trades_df['Net_Return']).cumprod()

# Calculate Drawdown
equity = trades_df['Equity_Curve']
running_max = equity.cummax()
drawdown = (equity - running_max) / running_max
max_drawdown = drawdown.min() * 100

print(f"Total Non-Overlapping Trades: {len(trades_df)}")
print(f"Net Cumulative Return       : {(equity.iloc[-1] - 1) * 100:.2f}%")
print(f"Average Net Return / Trade  : {trades_df['Net_Return'].mean() * 100:.3f}%")
print(f"Net Win Rate                : {(trades_df['Net_Return'] > 0).mean() * 100:.1f}%")
print(f"Maximum Drawdown            : {max_drawdown:.2f}%")
```
